```python
import os
import shutil
from google.colab import userdata

# --- Configuration Section ---
# Fetching your GITHUB_TOKEN securely from Colab Secrets
try:
    GITHUB_TOKEN = userdata.get('GITHUB_TOKEN')
except Exception as e:
    print("Error: Make sure you have added GITHUB_TOKEN to Colab Secrets (🔑 icon on the left sidebar) and enabled notebook access.")
    GITHUB_TOKEN = None

GITHUB_USER = "EmiBonClay" # Your GitHub Username
REPO_NAME = "orchidaceae_cl"     # Your Repository Name
SOURCE_DIR = "/content/drive/MyDrive/Gavilea"

CLONE_DIR = f"/content/{REPO_NAME}"
REPO_URL = f"https://{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{REPO_NAME}.git" if GITHUB_TOKEN else None
# ------------------------------

if GITHUB_TOKEN and GITHUB_USER != "YOUR_GITHUB_USERNAME":
    # 1. Clone the GitHub repository if it doesn't already exist
    if not os.path.exists(CLONE_DIR):
        print(f"Cloning {REPO_NAME}...")
        !git clone {REPO_URL} {CLONE_DIR}
    else:
        print(f"Repository already cloned at {CLONE_DIR}")

    # 2. Function to copy image files
    def copy_images(src, dest):
        image_extensions = ('.png', '.jpg', '.jpeg', '.gif', '.bmp', '.tiff', '.webp')
        copied_count = 0

        for root, dirs, files in os.walk(src):
            for file in files:
                if file.lower().endswith(image_extensions):
                    # Maintain subfolder structure in the destination
                    relative_path = os.path.relpath(root, src)
                    target_dir = os.path.join(dest, "images", relative_path)
                    os.makedirs(target_dir, exist_ok=True)

                    src_file_path = os.path.join(root, file)
                    dest_file_path = os.path.join(target_dir, file)

                    shutil.copy2(src_file_path, dest_file_path)
                    copied_count += 1
        print(f"Copied {copied_count} images to {dest}/images")

    # 3. Copy images from Google Drive to the cloned repo
    if os.path.exists(SOURCE_DIR):
        copy_images(SOURCE_DIR, CLONE_DIR)
    else:
        print(f"Error: Source directory {SOURCE_DIR} does not exist. Please check your Google Drive path.")

    # 4. Git Commit and Push changes with authenticated URL
    %cd {CLONE_DIR}
    !git config --global user.email "emibonclay@gmail.com"
    !git config --global user.name "{GITHUB_USER}"

    # Force update the remote URL to use the token authenticated URL
    !git remote set-url origin {REPO_URL}

    !git add .
    !git commit -m "Add images from Google Drive Gavilea folder" --allow-empty
    !git push origin main
else:
    print("Please configure GITHUB_TOKEN in Secrets and update GITHUB_USER/REPO_NAME variables in the code.")
```
